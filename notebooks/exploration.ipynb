# Speech Emotion Recognition — Data Exploration
Quick EDA: waveform, spectrogram, and MFCC visualization for a sample audio clip.

```python
import sys
sys.path.append('../src')
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np

# Point this to any sample RAVDESS wav file after you've downloaded the dataset
SAMPLE_FILE = '../data/RAVDESS/Actor_01/03-01-03-01-01-01-01.wav'

y, sr = librosa.load(SAMPLE_FILE, sr=22050)
```

```python
plt.figure(figsize=(10, 3))
librosa.display.waveshow(y, sr=sr)
plt.title('Waveform')
plt.show()
```

```python
mel = librosa.feature.melspectrogram(y=y, sr=sr, n_mels=128)
mel_db = librosa.power_to_db(mel, ref=np.max)
plt.figure(figsize=(10, 4))
librosa.display.specshow(mel_db, sr=sr, x_axis='time', y_axis='mel')
plt.colorbar(format='%+2.0f dB')
plt.title('Mel Spectrogram')
plt.show()
```

```python
mfcc = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=40)
plt.figure(figsize=(10, 4))
librosa.display.specshow(mfcc, sr=sr, x_axis='time')
plt.colorbar()
plt.title('MFCC')
plt.show()
```
